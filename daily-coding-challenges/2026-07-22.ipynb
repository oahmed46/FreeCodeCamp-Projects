Piggy Bank
Given an object representing a piggy bank, return the total value as a string formatted as "$D.CC".

The object may contain any of the following:

Coin	Value
pennies	$0.01
nickels	$0.05
dimes	$0.10
quarters	$0.25


```python
def piggy_bank(coins):

    if len(coins) == 0:
        return "$0.00"

    value = {"pennies": 0.01, "nickels": 0.05, "dimes": 0.1, "quarters": 0.25}
    total = float(0.00)
    for key in coins:
        total += coins[key] * value[key]
        print(key)

    print(total)
    return f"${round(total,2)}"
```

```python
piggy_bank({"pennies": 3, "nickels": 5, "dimes": 2, "quarters": 6}) # "$1.98"
```

```python
piggy_bank({"pennies": 1, "nickels": 1, "dimes": 1, "quarters": 1}) # "$0.41"
```

```python
piggy_bank({"nickels": 8, "dimes": 6, "quarters": 5}) # "$2.25"
```

```python
piggy_bank({}) # "$0.00"
```

```python
piggy_bank({"pennies": 146, "nickels": 11, "dimes": 0, "quarters": 19}) # "$6.76"
```
