Emoji Translator

Given a string of emojis, return the phrase using the following table:
Emoji 	Word
👶 	"baby"
🐱 	"cat"
🐕 	"dog"
🐟 	"fish"
🥵 	"hot"
🧊 	"ice"
🪨 	"rock"
🦈 	"shark"
🍲 	"soup"
⭐ 	"star"

Return the words separated by spaces.


```python
def get_emoji_phrase(s):
    word = []
    table = {
    "👶":"baby",
    "🐱":"cat",
    "🐕":"dog",
    "🐟":"fish",
    "🥵":"hot",
    "🧊":"ice",
    "🪨":"rock",
    "🦈":"shark",
    "🍲":"soup",
    "⭐":"star"}
    for emoji in s:
        word.append(table[emoji])
    return " ".join(word)
```

```python
get_emoji_phrase("🪨⭐") # "rock star"
```

```python
get_emoji_phrase("🥵🐕") # "hot dog"
```

```python
get_emoji_phrase("👶🦈") # "baby shark"
```

```python
get_emoji_phrase("⭐🐟") # "star fish"
```

```python
get_emoji_phrase("🧊🧊👶") # "ice ice baby"
```

```python
get_emoji_phrase("🐱🐟🍲") # "cat fish soup"
```
