Magic Square Solver

Given a 3x3 grid with one missing number (represented as 0), return the missing number that completes the magic square, or "impossible" if no valid number exists.

A magic square is a grid where every row, column, and diagonal adds up to the same number.


```python
def solve_magic_square(grid):

    missing = find_missing(grid)

    if missing[1] == 0:
        number = add_row(grid,1)
    else:
        number = add_row(grid,0)

    solution = number - add_row(grid, missing[0])

    grid[missing[0]][missing[1]] = solution

    if number == add_column(grid,missing[1]):
        return solution
    return "impossible"


def find_missing(grid):
    missing = (0,0)
    for x in range(len(grid)):
        for y in range(len(grid[x])):
            if grid[x][y] == 0:
                missing = (x,y)
    return missing

def add_row(grid,row):
    number = 0
    for i in range(len(grid[row])):
        number += grid[row][i]
    return number


def add_column(grid,column):
    number = 0
    for i in range(len(grid[column])):
        number += grid[i][column]
    return number


def add_diagonal(grid,position):
    number = 0
    for i in range(len(grid[column])):
        number += grid[i][column]
    return number
```

```python
solve_magic_square([[2, 7, 6], [9, 0, 1], [4, 3, 8]]) # 5
```

```python
solve_magic_square([[0, 14, 12], [18, 10, 2], [8, 6, 16]]) # 4
```
