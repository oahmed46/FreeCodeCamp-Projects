URL Query Parser
Given a URL that contains a query string, parse the query string into an object (or dictionary) of key-value pairs.

The query string begins after the "?",
each parameter is separated by "&",
each key/value pair is separated by "="
For example, given "https://example.com/search?name=Alice&age=30", return:

{
  "name": "Alice",
  "age": "30"
}
All values should be returned as strings.

```python
def parse_url_query(url):
    query_string = url.split('?', 1)[-1]

    pairs = query_string.split('&')

    result = {}

    for pair in pairs:
        if '=' in pair:
            key, value = pair.split('=', 1)
            result[key] = value

    return result
```

```python
parse_url_query("https://example.com/search?name=Alice&age=30") # {"name": "Alice", "age": "30"}
```

```python
parse_url_query("https://freecodecamp.org/learn?skill=programming&language=python") # {"skill": "programming", "language": "python"}
```

```python
parse_url_query("https://freecodecamp.org/items?category=books&sort=asc&page=2") # {"category": "books", "sort": "asc", "page": "2"}
```

```python
parse_url_query("https://example.com?redirect=freecodecamp.org/learn&when=now") # {"redirect": "freecodecamp.org/learn", "when": "now"}
```
