Pig Latin Converter
Given a string, convert it to Pig Latin using the following rules:

If a word begins with a vowel ("a", "e", "i", "o", or "u"), add "way" to the end. For example, "universe" converts to "universeway".
If a word begins with one or more consonants, move them to the end and add "ay". For example, "hello" converts to "ellohay".
Preserve the case of the first letter. For example, "Hello" converts to "Ellohay".

```python
def pig_latin(s):
    vowels = ["a","e","i","o","u"]
    words = s.split()
    for i in range(len(words)):
        word = words[i]
        iscapital = False
        if word[0].isupper():
            iscapital = True
        if word[0].lower() in vowels:
            words[i] += "way"
        else:
            consonants = ""
            while word[0] not in vowels:
                consonants += word[0]
                word = word[1:]
            words[i] = word + consonants + "ay"
        if iscapital:
            words[i] = words[i].capitalize()
    words = " ".join(words)
    return words
```

```python
pig_latin("universe") # "universeway"
```

```python
pig_latin("hello") # "ellohay"
```

```python
pig_latin("hello universe") # "ellohay universeway"
```

```python
pig_latin("Hello universe") # "Ellohay universeway"
```

```python
pig_latin("Pig Latin is fun") # "Igpay Atinlay isway unfay"
```

```python
pig_latin("The quick brown fox jumped over the lazy dog") # "Ethay uickqay ownbray oxfay umpedjay overway ethay azylay ogday"
```
