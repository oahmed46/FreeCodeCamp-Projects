Food Chain

Given an array of [predator, prey] pairs, return the food chain from the apex predator down to the bottom.

    The apex predator is the animal that is never prey to another animal.
    Return the chain as an array of strings.

```python
def get_food_chain(pairs):

    food_chain = {}
    apex_predator = []

    for pair in pairs:
        food_chain[pair[0]] = pair[1]


    for key in food_chain:
        if key not in food_chain.values():
            apex_predator.append(key)

    for i in range(len(food_chain)):
        apex_predator.append(food_chain[apex_predator[-1]])

    return apex_predator
```

```python
get_food_chain([["cat", "mouse"]]) # ["cat", "mouse"]
```

```python
get_food_chain([["wolf", "deer"], ["deer", "grass"]]) # ["wolf", "deer", "grass"]
```

```python
get_food_chain([["hawk", "snake"], ["snake", "frog"], ["frog", "fly"]]) # ["hawk", "snake", "frog", "fly"]
```

```python
get_food_chain([["rabbit", "grass"], ["fox", "rabbit"], ["eagle", "fox"]]) # ["eagle", "fox", "rabbit", "grass"]
```

```python
get_food_chain([["seal", "salmon"], ["herring", "shrimp"], ["orca", "seal"], ["shrimp", "plankton"], ["salmon", "herring"]]) # ["orca", "seal", "salmon", "herring", "shrimp", "plankton"]
```
