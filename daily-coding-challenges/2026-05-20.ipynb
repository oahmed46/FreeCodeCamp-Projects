String Zipper
Given two strings, return a new string that interleaves their characters one at a time. If one string is longer, append the remaining characters at the end.

Begin with the first character of the first string.

```python
def zip_strings(a, b):
    zipped = ""
    if len(a) == len(b):
        for i in range(len(a)):
            zipped += a[i] + b[i]
    if len(a) > len(b):
        for i in range(len(b)):
            zipped += a[i] + b[i]
        zipped += a[len(b):len(a)]
    if len(b) > len(a):
        for i in range(len(a)):
            zipped += a[i] + b[i]
        zipped += b[len(a):len(b)]
    return zipped
```

```python
zip_strings("abc", "123") # "a1b2c3"
```

```python
zip_strings("acegikmoqsuwy", "bdfhjlnprtvxz") # "abcdefghijklmnopqrstuvwxyz"
```

```python
zip_strings("day", "night") # "dnaiyght"
```

```python
zip_strings("python", "javascript") # "pjyatvhaosncript"
```

```python
zip_strings("feCdCm", "reoeap") # "freeCodeCamp"
```
