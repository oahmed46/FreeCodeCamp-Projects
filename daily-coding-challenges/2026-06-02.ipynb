Schema Validator Part 2
Given an object (JavaScript) or dictionary (Python), determine if it matches the following schema:

{
  username: string,
  posts: number,
  verified: boolean
}
Extra keys are allowed

```python
def is_valid_schema(obj):
    return "username" in obj.keys() and isinstance(obj["username"], str) and "posts" in obj.keys() and isinstance(obj["posts"], int) and "verified" in obj.keys() and isinstance(obj["verified"], bool)
```

```python
is_valid_schema({"username": "alice", "posts": 10, "verified": False}) # True
```

```python
is_valid_schema({"username": "carol", "posts": 15, "verified": True, "followers": 25}) # True
```

```python
is_valid_schema({"username": "frank", "posts": "21", "verified": True}) # False
```

```python
is_valid_schema({"username": "sam", "posts": 17, "verified": "false"}) # False
```

```python
is_valid_schema({"username": "bill", "verified": True}) # False
```

```python
is_valid_schema({"username": "fred", "verified": True}) # False
```

```python
is_valid_schema({"username": 5, "posts": 10, "verified": True}) # False
```
