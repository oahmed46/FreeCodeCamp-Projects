Array Chunks
Given an array and a chunk size, return the array split into sub-arrays of that size.

The last chunk may be smaller if the array doesn't divide evenly.

```python
def chunk_array(arr, size):
    return [arr[i:i + size] for i in range(0, len(arr), size)]
```

```python
chunk_array([1, 2, 3, 4, 5, 6], 3) # [[1, 2, 3], [4, 5, 6]]
```

```python
chunk_array([1, "two", 3, "four", 5, "six", 7, "eight"], 2) # [[1, "two"], [3, "four"], [5, "six"], [7, "eight"]]
```

```python
chunk_array([1, 2, 3, 4, 5], 3) # [[1, 2, 3], [4, 5]]
```

```python
chunk_array(["a", "b", "c", "d", "e"], 1) # [["a"], ["b"], ["c"], ["d"], ["e"]]
```

```python
chunk_array([1, 2, 3], 5) # [[1, 2, 3]]
```
