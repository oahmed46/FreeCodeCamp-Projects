Schema Validator Part 1
Given an object (JavaScript) or dictionary (Python), determine if it matches the following schema:

{
  username: string
}
Extra keys are allowed

```python
def is_valid_schema(obj):
    return "username" in obj.keys() and isinstance(obj["username"], str)
```

```python
is_valid_schema({"username": "bob"}) # True
```

```python
is_valid_schema({"username": "jen", "posts": 30}) # True
```

```python
is_valid_schema({"username": ""}) # True
```

```python
is_valid_schema({"username": 7}) # False
```

```python
is_valid_schema({"posts": 25}) # False
```
