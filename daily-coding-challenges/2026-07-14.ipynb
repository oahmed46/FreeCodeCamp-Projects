Pet Age Calculator
Given a pet type and age in human years, return the equivalent age in pet years using the following conversion table:

Pet	Multiplier
"dog"	7
"cat"	6
"rabbit"	8
"hamster"	30
"guinea pig"	12
"goldfish"	6
"bird"	5

```python
def pet_years(pet, age):
    pet_age_table = {"dog":7,
    "cat":6,
    "rabbit":8,
    "hamster":30,
    "guinea pig":12,
    "goldfish":6,
    "bird":5}
    return age*pet_age_table[pet]
```

```python
pet_years("dog", 5) # 35
```

```python
pet_years("cat", 9) # 54
```

```python
pet_years("rabbit", 3) # 24
```

```python
pet_years("hamster", 4) # 120
```

```python
pet_years("guinea pig", 5) # 60
```

```python
pet_years("goldfish", 2) # 12
```

```python
pet_years("bird", 1) # 5
```
