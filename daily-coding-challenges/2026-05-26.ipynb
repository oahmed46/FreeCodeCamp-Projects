Sum of Differences
Given an array of numbers, return the sum of the differences between each number and the one that follows it.

For example, given [1, 3, 4], return 3 (2 + 1).

```python
def sum_of_differences(arr):
    sod = 0
    for i in range(len(arr)):
        if i+1 <= len(arr)-1:
            sod += (arr[i+1] - arr[i])
    return sod
```

```python
sum_of_differences([1, 3, 4]) # 3
```

```python
sum_of_differences([5, -3, 3, 9, 10]) # 5
```

```python
sum_of_differences([9, 6, 15, -20, 33, 14, 25, 16, -7]) # -16
```

```python
sum_of_differences([50, 102, -46, 82, -49, 29, 71, 902, -237, 111, -61, 75]) # 25
```
