1337 Speak
Given a lowercase string, return it translated into leet speak by replacing the letters below with their leet substitutions:

Letter	Leet
a	4
e	3
g	9
i	1
l	1
o	0
s	5
t	7
Characters with no substitution are left unchanged.

```python
def make_leet(s):
    leet = {
        "a":"4",
        "e":"3",
        "g":"9",
        "i":"1",
        "l":"1",
        "o":"0",
        "s":"5",
        "t":"7"
    }

    new = ""

    for l in s:
        if l in leet.keys():
            new += leet[l]
        else:
            new += l
    return new
```

```python
make_leet("cool") # "c001"
```

```python
make_leet("leet") # "1337"
```

```python
make_leet("hacker") # "h4ck3r"
```

```python
make_leet("satellite") # "547311173"
```

```python
make_leet("abcdefghijklmnopqrstuvwxyz") # "4bcd3f9h1jk1mn0pqr57uvwxyz"
```
