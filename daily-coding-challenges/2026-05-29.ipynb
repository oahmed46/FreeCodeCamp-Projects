```python
import math
def get_wider_aspect_ratio(a, b):
    a = [int(n) for n in a.split("x")]
    b = [int(n) for n in b.split("x")]

    aspect_a = a[0]/a[1]
    aspect_b = b[0]/b[1]

    if aspect_a > aspect_b:
        gcd = math.gcd(a[0],a[1])
        return f"{a[0] // gcd}:{a[1] // gcd}"
    else:
        gcd = math.gcd(b[0],b[1])
        return f"{b[0] // gcd}:{b[1] // gcd}"
```

```python
get_wider_aspect_ratio("1920x1080", "800x600") # "16:9"
```

```python
get_wider_aspect_ratio("1080x1350", "2048x1536") # "4:3"
```

```python
get_wider_aspect_ratio("640x480", "2440x1220") # "2:1"
```

```python
get_wider_aspect_ratio("360x640", "1080x1920") # "9:16"
```
