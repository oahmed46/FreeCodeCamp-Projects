Golden Ratio
Given two numbers, determine if their ratio approximates the golden ratio.

Use a golden ratio of 1.618
Allow a tolerance of 0.01

```python
def is_golden_ratio(a, b):
    golden_ratio = 1.618
    tolerance = 0.01

    if a > b:
        ratio = a/b
    else:
        ratio = b/a

    if golden_ratio - tolerance < ratio < golden_ratio +tolerance:
        return True
    return False
```

```python
is_golden_ratio(21, 34) # True
```

```python
is_golden_ratio(15, 20) # False
```

```python
is_golden_ratio(8, 13) # True
```

```python
is_golden_ratio(10, 16) # False
```

```python
is_golden_ratio(1618, 1000) # True
```

```python
is_golden_ratio(88, 55) # False
```
