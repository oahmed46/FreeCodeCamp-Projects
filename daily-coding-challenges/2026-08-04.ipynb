Golf Handicap Calculator

Given an array of golf scores and a corresponding array of course par values, return the golfer's handicap index using the following method:

    Calculate the differential for each round by subtracting the par from the score, then return the average of all differentials rounded to one decimal place.



```python
from decimal import Decimal, ROUND_HALF_UP

def calculate_handicap(scores, pars):
    difference = []
    for i in range(len(scores)):
        difference.append(scores[i] - pars[i])
    average = 0
    for value in difference:
        average += value
    result = float(Decimal(str(average / len(difference))).quantize(Decimal('1.1'), rounding=ROUND_HALF_UP))
    return result
```

```python
calculate_handicap([72, 72, 72], [72, 72, 72]) # 0
```

```python
calculate_handicap([80, 76, 78, 78], [72, 72, 72, 72]) # 6
```

```python
calculate_handicap([42, 45, 46, 44], [36, 36, 36, 36]) # 8.3
```

```python
calculate_handicap([85, 80, 76, 79, 82], [72, 72, 72, 71, 71]) # 8.8
```

```python
calculate_handicap([41, 50, 48, 52, 46, 49], [35, 37, 35, 37, 35, 37]) # 11.7
```
