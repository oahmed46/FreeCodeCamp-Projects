Word Blender
Given two words, return a new word by combining the first half of the first word with the second half of the second word.

For odd-length words, the first half is the shorter half.

```python
def blend_words(word1, word2):
    return word1[:int(len(word1)/2)] + word2[int(len(word2)/2):]
```

```python
blend_words("turtle", "toucan") # "turcan"
```

```python
blend_words("chipmunk", "flamingo") # "chipingo"
```

```python
blend_words("falcon", "pelican") # "falican"
```

```python
blend_words("hyena", "iguana") # "hyana"
```

```python
blend_words("scorpion", "gorilla") # "scorilla"
```

```python
blend_words("platypus", "wolverine") # "platerine"
```
