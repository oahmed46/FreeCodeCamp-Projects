Character Frequency
Given a string, return an object (JavaScript) or dictionary (Python) mapping each character to the number of times it appears.

```python
def get_frequency(s):
    char_freq = {}
    for w in s:
        count = 0
        for i in range(len(s)):
            if w in s[i]:
                count += 1
        char_freq[w] = count
    return char_freq
```

```python
get_frequency("test") # {"t": 2, "e": 1, "s": 1}
```

```python

get_frequency("mississippi") # {"m": 1, "i": 4, "s": 4, "p": 2}
```
