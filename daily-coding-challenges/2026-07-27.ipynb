Pronic Number
Given a number, determine whether it is a pronic number.

A pronic number is the product of two consecutive integers. For example, 6 is pronic because 2 * 3 = 6.

```python
def is_pronic(n):
    if n == 0:
        return True
    for i in range(n):
        if i*(i+1) == n:
            return True
    return False
```

```python
is_pronic(6) # True
```

```python
is_pronic(15) # False
```

```python
is_pronic(12) # True
```

```python
is_pronic(132) # True
```

```python
is_pronic(80) # False
```

```python
is_pronic(0) # True
```
