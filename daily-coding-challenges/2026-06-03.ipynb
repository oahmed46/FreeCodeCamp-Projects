Schema Validator Part 3
Given an object (JavaScript) or dictionary (Python), determine if it matches the following schema:

Roles = "user" | "creator" | "moderator" | "staff" | "admin"

{
  username: string,
  posts: number,
  verified: boolean,
  role: Roles
}
The pipe (|) symbol means "or". role must be one of the listed Roles values.
Extra keys are allowed

```python
def is_valid_schema(obj):
    roles = ["user","creator","moderator","staff","admin"]
    if "username" not in obj:
        return False
    if not isinstance(obj["username"],str):
        return False
    if "posts" not in obj:
        return False
    if not isinstance(obj["posts"],int):
        return False
    if "verified" not in obj:
        return False
    if not isinstance(obj["verified"],bool):
        return False
    if "role" not in obj:
        return False
    if obj["role"] not in roles:
        return False

    return True
```

```python
is_valid_schema({"username": "henry", "posts": 0, "verified": True, "role": "staff"}) # True
```
