Word Score
Given a word, return its score using a standard letter-value table:

Letter	Value
A	1
B	2
...	...
Z	26
Upper and lowercase letters have the same value.

```python
def get_word_score(word):
    letter = [chr(x) for x in range(ord('a'), ord('z')+1)]
    value = list(range(1,28))
    word_score = dict(zip(letter,value))
    score = 0
    for w in word.lower():
        score += word_score[w]
    return score
```

```python
get_word_score("hi") # 17
```

```python
get_word_score("hello") # 52
```

```python
get_word_score("hippopotamus") # 169
```

```python
get_word_score("freeCodeCamp") # 94
```
